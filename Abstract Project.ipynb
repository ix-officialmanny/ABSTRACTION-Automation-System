## Case Study Activity: Building Automation System

UMaT's new auditorium contains different automated systems. Although each
system performs a different task, they must all support common operations.

**Requirements:**
- Create an abstract class: `BuildingSystem`
- Abstract methods: `start()`, `stop()`, `status()`
- Child classes: `AirConditioningSystem`, `LightingSystem`, `SecuritySystem`
- Each child class must implement all abstract methods.
- Store the objects in a list and demonstrate polymorphic behaviour by
  calling each method.
- Add a new class (`FireAlarmSystem`) without modifying the existing
  processing loop.

```python
from abc import ABC, abstractmethod


class BuildingSystem(ABC):
    """Abstract base class representing a generic building automation system."""

    def __init__(self, name):
        self.name = name

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

    @abstractmethod
    def status(self):
        pass


class AirConditioningSystem(BuildingSystem):
    def start(self):
        print(f"{self.name}: Cooling system activated.")

    def stop(self):
        print(f"{self.name}: Cooling system deactivated.")

    def status(self):
        print(f"{self.name}: Currently maintaining set temperature.")


class LightingSystem(BuildingSystem):
    def start(self):
        print(f"{self.name}: Lights turned ON.")

    def stop(self):
        print(f"{self.name}: Lights turned OFF.")

    def status(self):
        print(f"{self.name}: Lighting status is normal.")


class SecuritySystem(BuildingSystem):
    def start(self):
        print(f"{self.name}: Security monitoring activated.")

    def stop(self):
        print(f"{self.name}: Security monitoring deactivated.")

    def status(self):
        print(f"{self.name}: All entry points secure.")


# New class added WITHOUT modifying the existing processing loop below
class FireAlarmSystem(BuildingSystem):
    def start(self):
        print(f"{self.name}: Fire alarm system armed.")

    def stop(self):
        print(f"{self.name}: Fire alarm system disarmed.")

    def status(self):
        print(f"{self.name}: No fire detected. System nominal.")


# Store objects in a list and demonstrate polymorphic behaviour
building_systems = [
    AirConditioningSystem("AC Unit - Auditorium"),
    LightingSystem("Lighting - Auditorium"),
    SecuritySystem("Security - Auditorium"),
    FireAlarmSystem("Fire Alarm - Auditorium"),  # added seamlessly
]

for system in building_systems:
    system.start()
    system.status()
    system.stop()
    print("-" * 40)
```
